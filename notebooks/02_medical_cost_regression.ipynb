# Medical Cost Prediction Using Regression

## Research Question

How accurately can individual medical insurance charges be predicted using the available demographic and lifestyle characteristics?

## Hypothesis

Based on exploratory data analysis, smoking status, age, and BMI are expected to be the strongest predictors of medical insurance charges. Smoking status appears to have the strongest relationship with charges, while BMI may interact with smoking status.

Because the dataset contains a limited number of demographic and lifestyle characteristics, these variables may not explain all of the variation in individual medical costs.

## Target Variable

`charges` — individual medical insurance charges in USD.

## Predictor Variables

- `age`
- `sex`
- `bmi`
- `children`
- `smoker`
- `region`
